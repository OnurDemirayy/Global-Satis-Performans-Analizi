```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("sample_data.csv", encoding="ISO-8859-1", sep=";")  # Sütunları ; ile doğru ayırmak ve Türkçe karakterlerin hatasız okunması için birkaç parametre ekledim.
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])  #vEri setinde kaç tane eksik veri oldugunu gösterir.
```

```python
df.columns = df.columns.str.strip() # Sütün içindeki gizli boşlukları temizliyoruz.
```

```python
numeric_cols = ['Units Sold', 'Gross Sales', 'Discounts', 'Sales', 'COGS', 'Profit']
for col in numeric_cols:
    if col in df.columns:
        # Eğer veri metin olarak algılandıysa içindeki boşlukları ve karakterleri temizle
        df[col] = df[col].astype(str).str.replace('.', '').str.replace(',', '.')
        df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
for col in numeric_cols:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val) ## Satış ve kâr verileri birbirine bağlı olduğu için eksiklikleri medyan (orta değer) ile tamamlıyoruz
```

```python
categorical_cols = ['Segment', 'Country', 'Product', 'Discount Band']
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')  ## Kategori sütunlarında boşluk varsa 'Unknown' yaz
```

```python
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0])   #Tekrar verimize bakıyoruz ve temiz oldugunu görüyoruz.
```

```python
df['Date'] = df['Date'].ffill()  # Eksik tarih varsa bir önceki tarihten kopyalıyoruz
```

```python
print(f"Kalan eksik veri sayısı: {df.isnull().sum().sum()}") # Son kontrol
```

```python
# Temizlenmiş veriyi yeni bir dosya olarak kaydet
df.to_csv('clean_sample_data.csv', index=False, sep=';')
print("\n Veri başarıyla temizlendi ve 'clean_financial_data.csv' olarak kaydedildi!")
```
